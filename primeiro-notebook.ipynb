```python
import seaborn as sns
```

```python
#Colocamos a base de dados, o eixo x(Temperatura), kde é a linha de alteração, e bins é a quantidade de barrinhas.
grafico = sns.histplot(data=df, x="temperatura_c", kde=True, bins=20)
#Define um titulo
grafico.set_title("Distribuição de temperaturas")
#Define um titulo para a linha X
grafico.set_xlabel("Temperaturas em °C")
#Define o nome da linha Y
grafico.set_ylabel("Registros")
```
