```python
import seaborn as sns #Importei a biblioteca aonde irei criar os gráficos
```

```python
grafico_novo = sns.histplot(data=df, x="categoria", kde=True, bins=5) #Criei o gráfico de linha, onde o DF é a base de dados, X recebe os dados da coluna de "categoria", KDE é a linha de tendencia, e BINS é a quantidade de colunas
grafico_novo.set_title("Categorias mais vendidas") #Defini o título
grafico_novo.set_xlabel("Produtos") #Linha X
grafico_novo.set_ylabel("Quantidade") #Linha Y
```

```python
grafico_novo = sns.boxplot(data=df, x="categoria", y="preco_unitario", ) #Criei o gráfico boxplot (Mostra a mediana)
grafico_novo.set_title("Preço por Categoria") #Defini o título
grafico_novo.set_xlabel("Categoria") #Defini o nome da linha X
grafico_novo.set_ylabel("Preço Unitário") #Defini o nome da linha Y
```
